```python
#변수 선언 초기화해주기
money, c500, c100, c50, c10 = 0,0,0,0,0

money = int(input("교환할 돈 얼마인지"))

c500 = money // 500
money %= 500

c100 = money // 100
money %=100

c50 = money // 50
money%=50

c10 = money // 10
money %=10

print("\n 500원짜리 => %d개" % c500)
print(" 100원짜리 => %d개" % c100)
print(" 50원짜리 => %d개" % c50)
print(" 10원짜리 => %d개" % c10)
print(" 바꾸지 못한 잔 돈=> %d원 \n" % money)
```

```python
start = int(input("시작 연도를 입력하세요 : "))
end = int(input("끝 연도를 입력하세요 : "))

for year in range(start, end + 1):
    if ((year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)):
        print(f"{year}년은 윤년입니다.")
    else:
        print(f"{year}년은 윤년이 아닙니다.")
```
