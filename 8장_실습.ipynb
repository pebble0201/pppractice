```python
inStr = " 한글 Python 프로그래밍 "
outStr = ""

for i in range(0, len(inStr)) :
  if inStr[i] != ' ' :
    outStr += inStr[i]

print("원래 문자열 ==> " + '[' + inStr + ']')
print("공백 삭제 문자열 ==> " + outStr)
```

```python
inStr = "<<< 파<이>썬 >>>"
outStr = ""

for i in range(0, len(inStr)) :
  if inStr[i] != '<' and inStr[i] != '>' :
    outStr += inStr[i]

print("원래 문자열 ==> " + '[' + inStr + ']')
```

```python
str = input("문자열 입력 :")

if str.isdigit() :
  print("숫자임")
elif str.isalpha() :
  print("글자임")
elif str.isalnum() :
  print("글자 + 숫자")
else :
  print("모르겠")
```
