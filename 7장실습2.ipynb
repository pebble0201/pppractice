```python
foods = {"떡볶이":"어묵",
         "짜장면":"단무지",
         "라면":"김치",
         "삼겹살":"상추"};

while(True):
  myfood = input(str(list(foods.keys())) + "중 좋아하는 음식은?")
  if myfood in foods :
    print("<%s>궁합 음식은 <%s>입니다." %(myfood, foods.get(myfood)))
  elif myfood == "끝" :
    break
  else :
    print("그런음식없음")
```
